# Try My Voice

Runs the full app (teach 30 phrases, train your personal layer, talk) on a free Colab GPU and gives you a public link to open on your phone.

**How:** Runtime > Change runtime type > **T4 GPU**, then Runtime > **Run all**. Open the link it prints. Keep this tab open while you use the app.

Tip: run `01_train_base.ipynb` first in the same runtime to also use the shared model trained on TORGO dysarthric speech.
